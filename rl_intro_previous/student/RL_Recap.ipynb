# Three questions organize the methods

**Why this section?** There are many methods. Three questions place each one.

- **Is the model needed?**
    - *model-based*: a model of $\transp$ and the reward, known or learned, is used for planning (here: known, in value and policy iteration)
    - *model-free*: only sampled experience (everything from Monte Carlo on)
- **What does it learn?**
    - *values*: $\statevalfun$ or $\actvalfun$; the policy is read off them (the greedy action)
    - *a policy*: $\pi_\theta$ directly, by gradient ascent
    - *both*: actor-critic methods such as PPO learn a policy and a value (the critic)
- **Which policy generates the data?**
    - *on-policy*: the policy being learned (SARSA, REINFORCE); Monte Carlo and TD as taught here also evaluate the policy that produced the episodes; PPO and GRPO are *near* on-policy (a slightly older copy of the policy)
    - *off-policy*: another policy, such as an older or a more exploratory one (Q-learning, DQN)


**The methods of the series**

| Method | Model needed? | What it learns | Data comes from | Session |
|:--|:--|:--|:--|:-:|
| value iteration | yes (known) | state values $\statevalfun_*$ | none: planning | 2 |
| policy iteration | yes (known) | values of the current policy, then a better policy | none: planning | 2 |
| Monte Carlo | no | state values | complete episodes of the policy being evaluated | 3 |
| TD (one-step) | no | state values | the policy being evaluated, one step at a time | 3 |
| Q-learning | no | action values $\actvalfun$ | any behavior policy: off-policy | 3 |
| SARSA | no | action values | the policy itself: on-policy | 4 |
| DQN | no | action values (a network) | replay of older behavior: off-policy | 4 |
| REINFORCE | no | the policy | the policy itself: on-policy | 5-6 |
| PPO | no | the policy and a value (the critic) | the old policy's episodes, reused for a few epochs: near on-policy | 7 |
| GRPO | no | the policy (no critic) | the old policy's group of completions per prompt: near on-policy | 7-8 |
| RLHF / DPO (preview) | no | a policy tuned to preferences | RLHF: responses sampled from the policy, scored by a learned reward model (on-policy). DPO: a fixed dataset of comparisons (offline) | Advanced course |

- SFT (supervised fine-tuning) is not RL: it imitates example responses; RL, RLHF and DPO start from it
- policy iteration is value-based dynamic programming: it alternates evaluating a policy with improving it, and uses no gradient


# One equation per idea

**Why this slide?** The series in one table: each idea condensed to its equation.

| Session | The idea | The equation |
|:-:|:--|:--|
| tour | the return | $\rlret_\tt = \rewseq_{\tt+1} + \disc \, \rlret_{\tt+1}$ |
| 2 | the Bellman equation | $\statevalfun_\pi(\state) = \sum_\act \pi(\act \mid \state) \sum_{\state'} \transp(\state' \mid \state, \act) \bigl( \rew(\state, \act, \state') + \disc \, \statevalfun_\pi(\state') \bigr)$ |
| 3 | TD learning | $\statevalfun(\stateseq_\tt) \leftarrow \statevalfun(\stateseq_\tt) + \alpha \bigl[ \rewseq_{\tt+1} + \disc \, \statevalfun(\stateseq_{\tt+1}) - \statevalfun(\stateseq_\tt) \bigr]$ |
| 3 (4) | Q-learning | $\actvalfun(\stateseq_\tt, \actseq_\tt) \leftarrow \actvalfun(\stateseq_\tt, \actseq_\tt) + \alpha \bigl[ \rewseq_{\tt+1} + \disc \max{\act'} \actvalfun(\stateseq_{\tt+1}, \act') - \actvalfun(\stateseq_\tt, \actseq_\tt) \bigr]$ |
| 5 | policy gradient | $\nabla_\theta J(\theta) = \Exp{\tau \sim \pi_\theta} \bigl[ \sum_\tt \nabla_\theta \log \pi_\theta(\actseq_\tt \mid \stateseq_\tt) \, \rlret_{0,\tau} \bigr]$, with $J$ the expected return |
| 6 | the advantage | the return replaced by $\advseq_\tt$ (return minus a baseline): the same gradient on average (for a baseline that does not depend on the action), less noise; in practice the advantage is estimated with a critic and GAE |
| 7 | PPO | $\min{} \bigl( r_\tt(\theta) \hat{\advseq}_\tt, \; \text{clip}(r_\tt(\theta), 1 - \epsilon_{\text{clip}}, 1 + \epsilon_{\text{clip}}) \hat{\advseq}_\tt \bigr)$, with $r_\tt(\theta)$ the new/old probability ratio and $\hat{\advseq}$ an advantage estimate; only the clipped term, to which a value loss and (for language models) a KL penalty are added |
| 7-8 | GRPO | $\hat{\advseq}_i = (\rew_i - \text{mean}) / \text{std}$ over the $G$ responses to one prompt (the $G$ of group size, not the return); this replaces the critic |


# Choosing a family

**Why this slide?** From knowing the methods to choosing among them for a new problem.

- **The model is known and small** (a board game rule set, a grid): dynamic programming
- **Small discrete states and actions, the model unknown**: tabular Q-learning or SARSA
- **Large or continuous states, discrete actions**: DQN
- **Continuous actions, or a stochastic policy is needed**: policy gradient methods; PPO is the usual default
- **Language models**: the policy is the model; rewards come from a program, a reward model or a judge; PPO, GRPO and preference methods are the tools
- in every case: reward design decides what the agent learns (reward hacking), and exploration decides what *can* be learned (RL mostly amplifies what it already tries)
