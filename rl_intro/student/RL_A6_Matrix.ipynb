## Three patterns in the table

**Why this section?** The table is a map of trade-offs, not a list.

> **Three patterns: the width-by-depth dial, the bias/variance trade, and the on/off split.**

- **the width-by-depth dial**
    - dynamic programming: wide, shallow
    - Monte Carlo: narrow, to the end
    - TD: narrow and shallow, $p$-step between
    - depth moves two things together: how much of the target is our own estimate (bias) and how much is sampled (variance)
    - **width decides whether there is sampling variance at all**
        - a full backup averages over all successors exactly, so nothing is random and the target never fluctuates
        - a sampled backup uses one draw, so the target fluctuates from draw to draw
    - depth decides how much: each extra sampled step adds another random draw

**The bias/variance trade**

- looking to the end of the episode gives no bias and high variance
- one step gives lower variance and bias while the estimate is wrong
- the same trade returns in Part B, for the weight on each gradient term

**The on/off split**

- SARSA learns about the policy it follows
- Q-learning learns $\actvalfun_*$, the action values of an optimal policy $\pi_*$, while following another policy (one that keeps exploring)
- with a one-step target that split is what makes DQN's replay legal
- it returns in PPO as the probability ratio
- (the control rows read a policy off values, by argmax: the next slide)


## What the table does not have: a policy column

**Why this section?** The transition to Part B, seen on the table.

- every row learns or computes values
- the control rows improve a policy only by an argmax over values (policy iteration keeps an explicit policy, improved this way)
- no row learns the policy's own parameters by gradient
- the limitation, on the next slides: the argmax needs few discrete actions, and the greedy policy it gives is deterministic
    - the exploring behavior policies are not deterministic, but they are not what is learned
- Part B adds rows with a different entry under *Learns*: the policy itself
    - **policy gradient** means gradient ascent on the policy's parameters to raise the expected return
    - and new columns, where the methods differ: the three slots
        - the weight on each step
        - how far one update may move
        - where the baseline comes from
- the full table, with Part B and Part C rows, is in the recap

**Where Part A stops: the argmax needs a short list of actions**

> **A network of action values outputs one value per action, so the argmax needs a short list of discrete actions.**

- Microduck's action is $14$ continuous joint commands: there is no list to take a maximum over
- maximizing over a continuous or enormous action set means an optimization at every single step: possible with extra machinery, but costly

**Where Part A stops: the greedy policy is deterministic**

> **The greedy policy is deterministic, and exploration is added by hand.**

- in a fully observed problem the best policy can be deterministic, so this is no flaw there
- but when states look identical to the agent (a *partially observable* problem), the best policy can be random
    - likewise when an adversary exploits a predictable agent
    - example: rock-paper-scissors, a game, not a single-agent problem
- a softmax over $\actvalfun$ can randomize, but nothing trains it to be the *right* randomness
- exploration is added by hand ($\epsilon$-greedy), where a stochastic policy has randomness built in

**Where Part A stops: the policy is learned indirectly**

- the network is trained on a value-prediction error, not on the return we care about
- the policy is whatever the argmax of the predictions happens to be

**The next move: a network that is the policy**

> **For these cases, use the third row of the table "Three tables a network could replace": the network *is* the policy.**

- $\pi_\theta(\act \mid \state)$ outputs a *distribution* over actions (probabilities for a short list, or the mean and spread of a continuous action)
- train $\theta$ by gradient steps to raise the expected return directly: **policy gradient**, Part B
- Part A never built this network; Part B does
- training it means differentiating the expected return
- it is the same move as the last session, a network for a table, applied to the policy instead of $\actvalfun$
- it has its own cost: the training signal is noisy (high variance), which we will see is the main difficulty of Part B
