**Recall: what we use from the last two sessions**

> **A state value is an average of returns. TD(0) moves it toward one sampled target.**

- the return: $\rlret_\tt = \sum_{k \geq 0} \disc^k \, \rewseq_{\tt+k+1}$
- the state value of a policy $\pi$:
$$
\statevalfun_\pi(\state) = \Exp{\pi}\left[ \rlret_\tt \mid \stateseq_\tt = \state \right]
$$
- the TD(0) update, from one sample:
$$
\statevalfun(\stateseq_\tt) \leftarrow \statevalfun(\stateseq_\tt) + \alpha \bigl[ \rewseq_{\tt+1} + \disc\,\statevalfun(\stateseq_{\tt+1}) - \statevalfun(\stateseq_\tt) \bigr]
$$

# Evaluating a policy is not yet acting

**Why this section?** The last session produced $\statevalfun_\pi$ for a policy $\pi$ that was playing.

> **Two things stand between those values and a good policy:**
>
> 1. **acting on state values needs the model**
> 2. **the experience depends on the policy that plays**

- **Problem 1: acting on state values needs the model**
    - $\statevalfun(\state)$ says how good a *state* is, not which *action* leads to the good ones
    - to pick the best action in $\state$, score each action $\act$ by the expected reward plus discounted value of the state it leads to
        - an *expectation over the possible next states* under $\transp$
    - that expectation is a sum weighted by the model's probabilities: without the model it cannot be computed
- in symbols (the lookahead of the dynamic-programming session):
$$
\pi(\state) = \argmax{\act} \; \underbrace{\sum_{\state', \rew} \transp(\state', \rew \mid \state, \act)}_{\text{weights: the model}} \bigl(\rew + \disc\,\statevalfun(\state')\bigr)
$$
- each action $\act$ is scored by the sum; the best score wins
    - $\statevalfun$ alone is not enough: the weights $\transp$ are needed

**Problem 2: the experience depends on the policy that plays**
    - an action that is never tried never gets a value
    - whatever the agent does decides which states and rewards it sees

**The plan**

> **Learn the value of *actions* instead of states, and learn it from samples (as in the last session). Acting then needs no model, which removes problem 1.**

- careful: the Bellman equation for action values still contains the model's probabilities, so *computing* $\actvalfun$ with that equation needs the model (as in dynamic programming)
    - only a *sampled* update (Q-learning, or SARSA) avoids it
- choose actions so that every action keeps being tried: that is problem 2
- the question left open last session ("must the episodes come from the policy we evaluate?") is answered in the on-policy / off-policy section


## Action values

**Why this section?** Storing a value for each state-*action* pair removes the lookahead when *acting*: the best action is read off directly.

> **An action value $\actvalfun_\pi(\state, \act)$ is the expected return of taking action $\act$ in state $\state$ and following $\pi$ afterwards.**

$$
\actvalfun_{\pi}: \States \times \Actions \to \Reals
$$

- it maps a state/action pair $(\state, \act)$ to that expected return
- $\actvalfun(\state, \act)$ is that expectation, already averaged over the next states and stored in the table
- learning it from samples absorbs the model's probabilities into the numbers

**What action values give us**

> **With $\actvalfun$, the best action is simply $\argmax{\act} \actvalfun(\state, \act)$: no model is needed for that step.**

- acting greedily on $\actvalfun_\pi$ gives a policy at least as good as $\pi$: the *policy improvement* step of policy iteration, now without the lookahead
- $\actvalfun_*$ is the action value when the policy is optimal, and $\statevalfun_*(\state) = \max{\act}\,\actvalfun_*(\state, \act)$
- the next slide shows what each stored value averages over (the Bellman equation), and why learning it from samples is what removes the model


**The Bellman equations for the action value**

> **Two versions: the expectation version for a policy $\pi$ (SARSA will sample it) and the optimality version (Q-learning will sample it).**

- both are the **one-step form**, written for action values (the forms table in A1)
- the expectation version (for a given policy $\pi$, not necessarily greedy), the one SARSA will sample
$$
\actvalfun_\pi(\state, \act) = \sum_{\state', \rew} \transp(\state', \rew \mid \state, \act)\bigl(\rew + \disc \sum_{\act'} \pi(\act' \mid \state')\,\actvalfun_\pi(\state', \act')\bigr)
$$

**The optimality version**

- no policy appears here: the max takes the best action under the same $\actvalfun_*$, so a policy greedy on $\actvalfun_*$ is optimal
- the optimality version, which value iteration for $\actvalfun$ turns into an update (with the model), and which Q-learning will sample
$$
\actvalfun_*(\state, \act) = \sum_{\state', \rew} \transp(\state', \rew \mid \state, \act)\bigl(\rew + \disc \max{\act'} \actvalfun_*(\state', \act')\bigr)
$$

**What the sum is over**

- what the sum (the expectation) is **over**
    - the possible next states $\state'$ and rewards $\rew$ after taking $\act$ in $\state$
    - weighted by the model's probabilities $\transp(\state', \rew \mid \state, \act)$
    - in the first version also the next action $\act'$, weighted by $\pi$
    - it is an average of $\rew + \disc\,(\text{value of the next state-action})$ over what the environment might do
- evaluating the right-hand side takes $\transp$: this is the model

The real gain of the switch from $\statevalfun$ to $\actvalfun$
- with $\statevalfun$, the model is needed twice: in the Bellman backup that learns the values, and in the lookahead that picks an action
- with $\actvalfun$, the average over next states is stored in the table entry for $(\state, \act)$
    - so acting is an $\argmax{\act}$ over stored numbers, with no model
- and the table can be learned **without** the model: replace the sum by *one* observed $(\rew, \state')$ and move the entry toward it (Q-learning, below)
- so the improvement is not that $\argmax{\act}$ is easier to find, but that the model is no longer needed


## Greed is not (always) good

**Why this section?** Action values fix problem 1 but not problem 2: a value exists only for actions that were tried. How should the agent choose its actions while it learns?

Before learning
- all values are $0$, so every action ties for best
- a *greedy* agent (always the action with the best current value) breaks the tie the same way every time
    - say, always the first action: 'left', which just bumps into the wall
- greedy on the *true* $\actvalfun_*$ is optimal
- the trouble here is greedy on *current estimates*, which are still wrong
- the action chosen in the first episode may not be the optimal one, and a deterministic policy locks us into it


This is the *exploration / exploitation* trade-off
- choosing the action currently believed to be best is **exploitation** (the *greedy* choice)
- choosing a different action is **exploration** (a *non-greedy* choice)

In model-free methods we may be forced to take actions that are not best by the current values, in order to sample the environment better.

An **$\epsilon$-greedy** policy makes
- the greedy choice with probability $(1-\epsilon)$
- a random action with probability $\epsilon$ (which may also be the greedy one)
- usually $\epsilon$ starts large and decays as the values improve

### Exploring starts: choosing which *states* we start in

$\epsilon$-greedy chooses *actions*. It does not choose which *states* we visit
- an update can only improve the value of a state-action pair the agent actually reaches
- if every episode begins in the same start state, some states may be reached rarely or never, however much we explore actions once there
    - example: a robot that always starts in one corner of a large grid and explores by taking mostly random steps
        - it seldom gets to the far corner
        - so the values there stay near their initial guess
- **exploring starts**: begin some episodes in a state drawn broadly across the whole state space, not only the usual start state
    - every state gets direct experience
    - needs a simulator (or a setting) in which we are free to start an episode anywhere
    - a reset is all we need, not the transition probabilities, so learning stays model-free
- two kinds of exploration, then: $\epsilon$-greedy for the *actions* tried, exploring starts for the *states* reached


**Greedy is not the same as best**

> **Greedy means best on the current information. Best (optimal, marked with a star: $\actvalfun_*$, $\pi_*$) means best given the truth.**

- a policy greedy on the current estimates $\actvalfun$ is wrong wherever $\actvalfun$ is wrong
- a policy greedy on the *true* $\actvalfun_*$ is optimal: it is $\pi_*$
- the hope: as the estimates improve, the greedy policy becomes $\pi_*$

**When is that hope justified?**

- for tabular Q-learning, $\actvalfun \to \actvalfun_*$ under two conditions
    - every state-action pair keeps being tried (this is what exploration is for)
    - the step size shrinks suitably
- acting greedily alone gives no such guarantee
    - a greedy agent may never try the action that is truly best
- with a neural network (the next session) the guarantee is lost, and the hope is only a hope


## Q-learning: learning action values from samples

**Why Q-learning?** There is no new idea here: it applies the sampling idea of the last session to the equation that finds the best policy, with action values so that no model is needed to act.

> **Recall the one idea of the last session: replace an expectation by a sample.**

- a value is an average over what the environment could do
- a sample is the one outcome it did choose
- Monte Carlo sampled the *definition* (the whole return); TD sampled the *one-step* Bellman equation
- the Bellman equation for action values has two versions as well (value slides), and dynamic programming used both with the model

The optimal action value satisfies the Bellman optimality equation
$$
\actvalfun_*(\state, \act) = \E\left[ \rewseq_{\tt+1} + \disc \max{\act'} \actvalfun_*(\stateseq_{\tt+1}, \act') \mid \stateseq_\tt = \state, \actseq_\tt = \act \right]
$$

Two things that look alike, and differ
- the **expectation** $\E[\cdot]$ averages over *all* the pairs $(\rewseq_{\tt+1}, \stateseq_{\tt+1})$ the environment *could* choose, each weighted by its probability under $\transp$
- the **sample** in the update below is the *single* pair the environment *did* choose this time
- one sample is a random draw from that same distribution: noisy, but right on average
    - so the update moves only part of the way toward it (step size $\alpha$)
    - many small moves average out to the expectation


**The Q-learning update**

> **Q-learning replaces the expectation by *one sample* and moves the estimate toward it.**
$$
\actvalfun(\stateseq_\tt, \actseq_\tt) \leftarrow \actvalfun(\stateseq_\tt, \actseq_\tt) + \alpha \left[ \rewseq_{\tt+1} + \disc \max{\act'} \actvalfun(\stateseq_{\tt+1}, \act') - \actvalfun(\stateseq_\tt, \actseq_\tt) \right]
$$

**Reading the Q-learning update**

- the bracket is a TD error, now for action values: the **target** (recall: the number the update moves the estimate toward) minus the current estimate
    - here the target is one sampled reward plus the discounted best value of the next state: a TD target
        - so it has the bias and variance profile of TD
        - biased while the table is wrong, low variance
- the target uses the best action in the next state, whichever action the agent actually takes next
    - with $\epsilon$-greedy the agent sometimes takes another (exploratory) action
        - so the action the target assumes can differ from the action taken
        - that gap is what *off-policy* means, named two sections on
- the policy is read off the table: $\pi(\state) = \argmax{\act} \actvalfun(\state, \act)$, the *greedy* policy of the current $\actvalfun$


A Q-learning update by hand, on (non-slippery) Frozen Lake
- all values start at $0$
- $\alpha = 0.9$, $\disc = 0.9$
    - a step size this close to $1$ is safe here only because nothing is random
        - with random transitions or rewards, each update would mostly replace the old estimate with the latest noisy sample
- the agent moves from square $14$ into **G** and earns $+1$
    - target $= 1 + 0.9 \times 0 = 1$ (the next state is terminal)
    - $\actvalfun(14, \text{right}) = 0 + 0.9\,(1 - 0) = 0.9$
- in a later episode it moves from square $13$ to square $14$ (reward $0$)
    - target $= 0 + 0.9 \times \max{\act'} \actvalfun(14, \act') = 0.9 \times 0.9 = 0.81$
    - $\actvalfun(13, \text{right}) = 0 + 0.9\,(0.81 - 0) \approx 0.73$
- information moves back one square each time a pair is revisited


**Pseudo code for Q-learning** (the book's own listing, in its notation: Sutton and Barto section 6.5; Sarsa is in section 6.4)

- initialize $\actvalfun(\state, \act)$ arbitrarily for all states and actions (often $0$)
- set the learning rate $\alpha$, the discount $\disc$ and the exploration rate $\epsilon$
- for each episode
    - initialize the state $\state$
    - repeat until $\state$ is terminal
        - with probability $\epsilon$ choose a random action $\act$ (exploration)
        - otherwise choose $\act = \argmax{\act'} \actvalfun(\state, \act')$ (exploitation)
        - take action $\act$
        - observe the reward $\rew$ and the next state $\state'$
        - $\actvalfun(\state, \act) \leftarrow \actvalfun(\state, \act) + \alpha \bigl[ \rew + \disc \max{\act'} \actvalfun(\state', \act') - \actvalfun(\state, \act) \bigr]$ (if $\state'$ is terminal, the max term is $0$)
        - $\state \leftarrow \state'$


**Notes on the pseudo-code**

- the exploration rate $\epsilon$ is shown as a constant
- in practice it usually decays over the episodes
- the table $Q$ is an estimate of $\actvalfun_*$
- the algorithm never uses $\transp$ or any reward other than the one received
- the action $a$ comes from the $\epsilon$-greedy rule
- the *update* assumes the next choices are greedy (the $\max{}$)

# Two policies in one update: on-policy and off-policy

**Why this section?** Last step we kept one question in mind: do the episodes have to come from the policy whose values we learn? Q-learning already says no.

Naming the two policies makes the answer precise, and lets us compare it with its smallest variant, SARSA.

> **Every update involves two policies: the behavior policy, which generates the data, and the target policy, which the update learns about.**

$$
\actvalfun(\state, \act) \leftarrow \actvalfun(\state, \act) + \alpha \Bigl[ \rew + \disc \, \underbrace{\max{\act'} \actvalfun(\state', \act')}_{\text{best next action}} - \actvalfun(\state, \act) \Bigr]
$$

Every update involves two policies
- the **behavior policy** picks the actions that are *actually taken*, so it generates the data
    - in Q-learning: $\epsilon$-greedy
    - it chose the action $\act$ whose value is updated
- the **target policy** is the policy whose value the update *learns about*
    - in Q-learning: the *greedy* policy: the $\max{\act'}$ always evaluates the best next action, whether or not the agent takes it

**What the update never uses**

- the next action the agent really takes in $\state'$ (again $\epsilon$-greedy, possibly a random one) is *never used* in the update


When the behavior and target policies are identical
- the method is **on-policy**

If the behavior and target policies can differ
- the method is **off-policy**

**The answer to last session's question is no**: Q-learning is **off-policy**
> **Q-learning learns $\actvalfun_*$, the action values of an optimal policy $\pi_*$, while following another policy ($\epsilon$-greedy) that keeps exploring.**

- the data come from one policy ($\epsilon$-greedy) and the update learns about another: greedy on the current $\actvalfun$
    - as $\actvalfun \to \actvalfun_*$, that greedy policy becomes $\pi_*$
- this holds even for the plain table: it needs no replay memory (reusing stored transitions comes next session)
- Monte Carlo and TD(0) as presented last session are on-policy: they evaluate the policy that played

Off-policy is about *which policy is learned*, not about model-free versus model-based.


**Aside (for your own review): why Q-learning needs no correction for the difference between the policies**

- $\actvalfun(\state, \act)$ conditions on the *first* action, which is fixed by the data
- the $\max{}$ then handles every later action by *definition*, not by sampling
- so no importance weights (re-weighting by how likely each policy was to take the action) are needed in the one-step case
- methods that sample the target policy's actions, and that reuse data from an older policy, do need such a ratio: it appears again in the policy gradient part


# SARSA: the on-policy counterpart

**Why this section?** With the two policies named, the smallest change to Q-learning is to learn about the policy that is actually acting. That is SARSA.

> **SARSA is a small change to the Q-learning update: the target uses the action actually taken next.**

- Q-learning target: $\rewseq_{\tt+1} + \disc \max{\act'} \actvalfun(\stateseq_{\tt+1}, \act')$
- SARSA target: $\rewseq_{\tt+1} + \disc \, \actvalfun(\stateseq_{\tt+1}, \actseq_{\tt+1})$
    - $\actseq_{\tt+1}$ is the action the behavior policy *actually chooses* next
$$
\actvalfun(\stateseq_\tt, \actseq_\tt) \leftarrow \actvalfun(\stateseq_\tt, \actseq_\tt) + \alpha \bigl[ \rewseq_{\tt+1} + \disc \, \actvalfun(\stateseq_{\tt+1}, \actseq_{\tt+1}) - \actvalfun(\stateseq_\tt, \actseq_\tt) \bigr]
$$


> **SARSA is the *sampled* version of the Bellman equation for $\actvalfun_\pi$**, just as TD(0) was for $\statevalfun_\pi$ (the expectation version is in the action-values section)

The name comes from the quintuple the update uses: $\stateseq_\tt, \actseq_\tt, \rewseq_{\tt+1}, \stateseq_{\tt+1}, \actseq_{\tt+1}$
- the exploratory choices are *inside* the values SARSA learns
- so it learns the value of the policy it follows: *on-policy*

        Choose a from s with the epsilon-greedy policy
        Repeat for each step:
            Take a, observe r and s'
            Choose a' from s' with the epsilon-greedy policy
            Q(s, a) = Q(s, a) + alpha * [r + gamma * Q(s', a') - Q(s, a)]     (Q = 0 at terminal states)
            s = s'; a = a'


## Grid-world: SARSA vs Q-learning on the cliff

- to goal cell $G$ (bottom right)
- the cliff cells fill the bottom row between them
- avoiding the cliff cells: stepping on one costs $-100$ (instead of $-1$), sends the agent back to $S$, and the episode continues

Both learners use tables (no neural network).

There is a negative reward ($-1$) for each time step so the reward is maximized by getting to the goal quickly.

There are four moves (up, right, down, left). The episode ends at $G$. Both learners explore $\epsilon$-greedily.


One transition, two targets (illustrative numbers)

Q-learning and SARSA therefore learn *different things*
- Q-learning learns the value of the **greedy** policy, even though it acts $\epsilon$-greedily
- SARSA learns the value of the **$\epsilon$-greedy** policy it actually follows, exploratory mistakes included

When exploration is harmless, both end up with similar values.
When exploration is *costly* (a cliff), the values differ, and so do the paths.

With a *fixed* $\epsilon$, SARSA's table is the value of the $\epsilon$-greedy policy, not $\actvalfun_*$ (it reaches $\actvalfun_*$ only if $\epsilon$ also decays: see the aside)

The cliff contrast below is for a *fixed* $\epsilon$: exploration stays on.


<img src="images/cliff_sarsa_vs_qlearning.png" width=60%>

- left: Q-learning's greedy path is the 13-step edge route
- SARSA's is usually a 17-step detour away from the edge (details in the aside)

**The cliff figure: the right panel and the setting**

- right: while exploring ($\epsilon = 0.1$), Q-learning earns *less* reward per episode, because it falls off the cliff
- setting: $\alpha = 0.5$, $\disc = 1$, $500$ episodes, $100$ runs, fixed $\epsilon = 0.1$


Why the paths differ
- SARSA's target uses the next action the behavior policy actually takes
    - near the cliff an exploratory step can fall off, and the large negative reward lowers the values that lead there
    - the lowered values spread to the states along the edge, so SARSA's greedy path moves away from the cliff
- Q-learning's target uses the max over the next actions
    - it ignores the exploratory step in the target, so the edge keeps its high value
    - its greedy path is the best one for an agent that stops exploring, but the exploring agent falls more often

Neither is wrong: they answer different questions
- Q-learning: what is the best path if I stop exploring?
- SARSA: what is the best path given that I keep exploring?


**Aside (for your own review): convergence conditions, cliff details and an animation**

- in the tabular case, Q-learning converges to $\actvalfun_*$ for any fixed $\epsilon > 0$, if every state-action pair keeps being visited and the step size decays suitably
- SARSA converges to $\actvalfun_*$ only if $\epsilon \to 0$ as well
- the cliff statistics: Q-learning's greedy path is the 13-step edge route in all 100 runs; SARSA's has 17 steps in 77 of 100 runs (the others learn a 19-step path or a route that is not finished)
- animation of the greedy paths after training: Q-learning (orange) hugs the cliff, SARSA (blue) goes around; settings $\alpha = 0.2$, $\disc = 0.99$, $\epsilon$ decaying to $0.01$, $5000$ episodes; it loops

# What we established, and what it sets up

- **Established**
    - with a model, state values give a policy by lookahead
    - without one, we learn *action values* and read the policy off them
    - an agent must try actions: $\epsilon$-greedy explores with probability $\epsilon$
    - and it must reach states: exploring starts begin episodes anywhere when a simulator allows
    - Q-learning is the sampled value iteration for $\actvalfun$: it learns $\actvalfun_*$ (the values of an optimal policy $\pi_*$) while acting $\epsilon$-greedily
    - every update has a **behavior policy** (generates the data) and a **target policy** (is learned about): equal means **on-policy** (SARSA), different means **off-policy** (Q-learning)
    - the two learn different things
    - on a cliff, different paths
- **Sets up**
    - **for the assignment:** a supplement applies Q-learning to pricing an American option
        - from a tree of prices down to a table
    - the next session starts from the limitation of a table
        - Microduck's state is $61$ continuous numbers, far too many to visit or store
    - a neural network learns $\actvalfun$ (deep Q-learning)
        - the same move could be applied to the policy itself
    - *replay* is legal *because* Q-learning is off-policy
        - replay: storing past transitions and reusing them for updates
    - after that: the argmax over actions does not handle continuous or stochastic actions, so we learn the policy directly (policy gradient)
    - bias and variance of the targets return there
